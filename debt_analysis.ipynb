```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

print(df.columns)
```

```python
df_melted = pd.melt(
    df,
    id_vars=[
        'country_name',
        'country_code',
        'series_name',
        'series_code'
    ],
    var_name='year',
    value_name='debt'
)
```

```python
df_melted.head()
```

```python
df_melted['debt'] = pd.to_numeric(
    df_melted['debt'],
    errors='coerce'
)
```

```python
df_melted.dropna(subset=['debt'], inplace=True)
```

```python
df_melted['year'] = pd.to_numeric(
    df_melted['year'],
    errors='coerce'
)
```

```python
df_melted.info()
```

# Exploratory Data Analysis (EDA)

```python
df_melted.describe()
```

```python
total_debt = df_melted['debt'].sum()

print("Total Global Debt:", total_debt)
```

```python
total_countries = df_melted['country_name'].nunique()

print("Total Countries:", total_countries)
```

```python
total_indicators = df_melted['series_name'].nunique()

print("Total Indicators:", total_indicators)
```

```python
top_countries = (
    df_melted.groupby('country_name')['debt']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_countries
```

```python
plt.figure(figsize=(14,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    hue=top_countries.index,   # FIX
    palette="viridis",
    legend=False               # removes extra legend
)

plt.title("Top 10 Countries by Total Debt")
plt.xlabel("Debt")
plt.ylabel("Country")

plt.show()
```

```python
lowest_countries = (
    df_melted.groupby('country_name')['debt']
    .sum()
    .sort_values()
    .head(10)
)

lowest_countries
```

```python
top_indicators = (
    df_melted.groupby('series_name')['debt']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_indicators
```

```python
plt.figure(figsize=(14,7))

sns.barplot(
    x=top_indicators.values,
    y=top_indicators.index,
    palette='magma'
)

plt.title("Top Debt Indicators")
plt.xlabel("Debt")
plt.ylabel("Indicator")

plt.show()
```

```python
yearly_debt = (
    df_melted.groupby('year')['debt']
    .sum()
)

yearly_debt.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    yearly_debt.index,
    yearly_debt.values,
    marker='o'
)

plt.title("Global Debt Trend Over Years")
plt.xlabel("Year")
plt.ylabel("Debt")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.histplot(
    df_melted['debt'],
    bins=50,
    kde=True
)

plt.title("Debt Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,4))

sns.boxplot(x=df_melted['debt'])

plt.title("Debt Outliers")

plt.show()
```
